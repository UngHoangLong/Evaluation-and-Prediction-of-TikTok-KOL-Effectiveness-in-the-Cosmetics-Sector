```python
def cosine_similarity(a, b):
    dot_product = sum(x * y for x, y in zip(a, b))
    magnitude_a = sum(x * x for x in a) ** 0.5
    magnitude_b = sum(x * y for x, y in zip(b, b)) ** 0.5
    return dot_product / (magnitude_a * magnitude_b) if magnitude_a * magnitude_b != 0 else 0

# Training data
training_data = [
    [19, 2, 1], [23, 6, 0], [24, 9, 0], [30, 1, 0], [16, 0, 1], [18, 1, 1], [22, 3, 1], [18, 1, 0], [15, 9, 1]
]
labels = ["Ha trang", "Ha trang", "Thu cuoi", "Thu cuoi", "Ha trang", "Thu cuoi", "Thu cuoi", "Ha trang", "Ha trang"]

# New data to predict
new_data = [[25, 9, 1], [35, 1, 0]]

# Predict labels for new data
for data in new_data:
    similarities = [cosine_similarity(data, train) for train in training_data]
    print(similarities)
    max_similarity_idx = similarities.index(max(similarities))
    
    predicted_label = labels[max_similarity_idx]
    print(f"Tuoi: {data[0]}, Gioi tinh: {data[1]} -> Bai hat: {predicted_label}")
```

```python
import math

# Data points: [number_of_posts, interaction_time]
data = [
    [8, 90], [15, 80], [2, 20], [3, 30], [3, 200],
    [6, 70], [1, 10], [4, 50], [8, 85], [2, 35]
]

# K-means implementation
def kmeans(data, k, max_iters=100):
    # Initialize centroids randomly from data points
    import random
    centroids = random.sample(data, k)
    
    for _ in range(max_iters):
        # Assign points to nearest centroid
        clusters = [[] for _ in range(k)]
        for point in data:
            distances = [math.sqrt(sum((x - c) ** 2 for x, c in zip(point, centroid))) for centroid in centroids]
            cluster_idx = distances.index(min(distances))
            clusters[cluster_idx].append(point)
        
        # Update centroids
        new_centroids = []
        for cluster in clusters:
            if cluster:  # Avoid division by zero
                new_centroid = [sum(d[i] for d in cluster) / len(cluster) for i in range(len(cluster[0]))]
                new_centroids.append(new_centroid)
            else:
                new_centroids.append(centroids[clusters.index(cluster)])
        
        # Check convergence
        if all(math.sqrt(sum((a - b) ** 2 for a, b in zip(c1, c2))) < 0.001 
               for c1, c2 in zip(centroids, new_centroids)):
            break
        centroids = new_centroids
    
    return clusters, centroids

# Run K-means with k=2 (you can adjust k)
k = 2
clusters, final_centroids = kmeans(data, k)

# Output results
for i, cluster in enumerate(clusters):
    print(f"Cluster {i + 1}: {cluster}")
print(f"Centroids: {final_centroids}")
```
